## Bureau

```python
bureau

# Need to make a singuler porfile per each SK_ID_CURR. 
# categorical variables convert to binaries, and numerical to mean, max, min 
# depending each context
```

```python
bureau_fe = bureau.copy()

#  Variables derivadas / ratios financieros

# Ratio deuda / crédito total:
# mide qué proporción del crédito otorgado sigue pendiente de pago
# Valores altos pueden indicar mayor carga financiera o exposición al riesgo
bureau_fe['DEBT_RATIO'] = np.where(
    bureau_fe['AMT_CREDIT_SUM'] > 0,
    bureau_fe['AMT_CREDIT_SUM_DEBT'] / bureau_fe['AMT_CREDIT_SUM'],
    np.nan
)

# Ratio mora / crédito total:
# mide qué proporción del crédito total está vencida
# Ayuda a capturar severidad de incumplimiento, no solo si existe mora
bureau_fe['OVERDUE_RATIO'] = np.where(
    bureau_fe['AMT_CREDIT_SUM'] > 0,
    bureau_fe['AMT_CREDIT_SUM_OVERDUE'] / bureau_fe['AMT_CREDIT_SUM'],
    np.nan
)

# Uso del límite de crédito:
# mide qué tanto del límite disponible está siendo utilizado como deuda
# Es útil especialmente en créditos tipo tarjeta o líneas de crédito
bureau_fe['LIMIT_USAGE'] = np.where(
    bureau_fe['AMT_CREDIT_SUM_LIMIT'] > 0,
    bureau_fe['AMT_CREDIT_SUM_DEBT'] / bureau_fe['AMT_CREDIT_SUM_LIMIT'],
    np.nan
)

# Ratio anualidad / crédito total:
# aproxima qué tan pesada es la cuota anual respecto al monto del crédito
# Valores altos pueden reflejar mayor presión de pago para el cliente
bureau_fe['ANNUITY_CREDIT_RATIO'] = np.where(
    bureau_fe['AMT_CREDIT_SUM'] > 0,
    bureau_fe['AMT_ANNUITY'] / bureau_fe['AMT_CREDIT_SUM'],
    np.nan
)

# Indicador de mora:
# identifica si el crédito tiene algún monto vencido
# Luego puede agregarse por cliente para obtener cantidad o proporción de créditos con mora
bureau_fe['HAS_OVERDUE'] = (
    bureau_fe['AMT_CREDIT_SUM_OVERDUE'] > 0
).astype(int)

# Indicador de deuda vigente:
# identifica si el cliente todavía tiene deuda pendiente en ese crédito
# Sirve para medir exposición activa al riesgo.
bureau_fe['HAS_DEBT'] = (
    bureau_fe['AMT_CREDIT_SUM_DEBT'] > 0
).astype(int)

# Indicador de crédito activo:
# marca si el crédito sigue abierto
# Permite calcular cuántos o qué proporción de créditos activos tiene cada cliente
bureau_fe['CREDIT_IS_ACTIVE'] = (
    bureau_fe['CREDIT_ACTIVE'] == 'Active'
).astype(int)

# Indicador de crédito cerrado:
# marca si el crédito ya fue cerrado.
# Una alta proporción de créditos cerrados puede reflejar historial crediticio completado.
bureau_fe['CREDIT_IS_CLOSED'] = (
    bureau_fe['CREDIT_ACTIVE'] == 'Closed'
).astype(int)


# Creamos dummies de variables categóricas
bureau_fe = pd.get_dummies(
    bureau_fe,
    columns=['CREDIT_ACTIVE', 'CREDIT_TYPE'],
    dtype=int
)


# Hacemos la agregación por cliente
bureau_summary = (
    bureau_fe
    .groupby('SK_ID_CURR')
    .agg({

        # Identificadores / volumen
        'SK_ID_BUREAU': ['count'],

        # Tiempo
        'DAYS_CREDIT': ['min', 'max', 'mean'],
        'CREDIT_DAY_OVERDUE': ['max', 'mean', 'sum'],
        'DAYS_CREDIT_ENDDATE': ['min', 'max', 'mean'],
        'DAYS_ENDDATE_FACT': ['min', 'max', 'mean'],
        'DAYS_CREDIT_UPDATE': ['min', 'max', 'mean'],

        # Montos de crédito
        'AMT_CREDIT_SUM': ['min', 'max', 'mean', 'sum'],
        'AMT_CREDIT_SUM_DEBT': ['min', 'max', 'mean', 'sum'],
        'AMT_CREDIT_SUM_LIMIT': ['min', 'max', 'mean', 'sum'],
        'AMT_CREDIT_SUM_OVERDUE': ['max', 'mean', 'sum'],
        'AMT_CREDIT_MAX_OVERDUE': ['max', 'mean'],

        # Anualidad
        'AMT_ANNUITY': ['max', 'mean', 'sum'],

        # Prolongaciones
        'CNT_CREDIT_PROLONG': ['max', 'mean', 'sum'],

        # Ratios
        'DEBT_RATIO': ['max', 'mean'],
        'OVERDUE_RATIO': ['max', 'mean'],
        'LIMIT_USAGE': ['max', 'mean'],
        'ANNUITY_CREDIT_RATIO': ['max', 'mean'],

        # Indicadores
        'HAS_OVERDUE': ['mean', 'sum'],
        'HAS_DEBT': ['mean', 'sum'],
        'CREDIT_IS_ACTIVE': ['mean', 'sum'],
        'CREDIT_IS_CLOSED': ['mean', 'sum'],

        # Dummies CREDIT_ACTIVE
        'CREDIT_ACTIVE_Active': ['mean'],
        'CREDIT_ACTIVE_Bad debt': ['mean'],
        'CREDIT_ACTIVE_Closed': ['mean'],
        'CREDIT_ACTIVE_Sold': ['mean'],

        # Dummies CREDIT_TYPE
        'CREDIT_TYPE_Another type of loan': ['mean'],
        'CREDIT_TYPE_Car loan': ['mean'],
        'CREDIT_TYPE_Cash loan (non-earmarked)': ['mean'],
        'CREDIT_TYPE_Consumer credit': ['mean'],
        'CREDIT_TYPE_Credit card': ['mean'],
        'CREDIT_TYPE_Interbank credit': ['mean'],
        'CREDIT_TYPE_Loan for business development': ['mean'],
        'CREDIT_TYPE_Loan for purchase of shares (margin lending)': ['mean'],
        'CREDIT_TYPE_Loan for the purchase of equipment': ['mean'],
        'CREDIT_TYPE_Loan for working capital replenishment': ['mean'],
        'CREDIT_TYPE_Microloan': ['mean'],
        'CREDIT_TYPE_Mobile operator loan': ['mean'],
        'CREDIT_TYPE_Mortgage': ['mean'],
        'CREDIT_TYPE_Real estate loan': ['mean'],
        'CREDIT_TYPE_Unknown type of loan': ['mean']
    })
)

# Modificamos nombres de columnas
bureau_summary.columns = [
    f'BUREAU_{col[0]}_{col[1]}'
    for col in bureau_summary.columns
]

bureau_summary = bureau_summary.reset_index()

#limpiamos infinitos y nulos derivados
bureau_summary = bureau_summary.replace([np.inf, -np.inf], np.nan)
```

## Installments_payments

```python
# Copia del dataframe
installments_fe = installments_payments.copy()

# Diferencia entre pago realizado y pago esperado:
# mide cuánto pagó de más o de menos respecto a la cuota programada
# Valores negativos indican subpago; positivos indican sobrepago
installments_fe['PAYMENT_DIFF'] = (
    installments_fe['AMT_PAYMENT']
    - installments_fe['AMT_INSTALMENT']
)

# Ratio de pago:
# mide qué proporción de la cuota esperada fue realmente pagada
# Valores:
#   = 1  -> pago exacto
#   < 1  -> pago incompleto
#   > 1  -> sobrepago
installments_fe['PAYMENT_RATIO'] = np.where(
    installments_fe['AMT_INSTALMENT'] > 0,
    installments_fe['AMT_PAYMENT']
    / installments_fe['AMT_INSTALMENT'],
    np.nan
)

# Days Past Due (DPD):
# mide cuántos días después de la fecha límite se realizó el pago
# Es una de las variables más importantes en riesgo crediticio
# Valores altos suelen asociarse con mayor probabilidad de default
installments_fe['DPD'] = (
    installments_fe['DAYS_ENTRY_PAYMENT']
    - installments_fe['DAYS_INSTALMENT']
)

# Days Before Due (DBD):
# mide cuántos días antes de la fecha límite se realizó el pago
# Captura comportamiento preventivo o disciplina financiera
installments_fe['DBD'] = (
    installments_fe['DAYS_INSTALMENT']
    - installments_fe['DAYS_ENTRY_PAYMENT']
)

# Eliminamos negativos en DPD:
# si el valor es negativo significa que pagó antes
# por lo tanto no existe retraso
installments_fe['DPD'] = (
    installments_fe['DPD']
    .apply(lambda x: max(x, 0))
)

# Eliminamos negativos en DBD:
# si el valor es negativo significa que pagó tarde
# por lo tanto no existe adelanto
installments_fe['DBD'] = (
    installments_fe['DBD']
    .apply(lambda x: max(x, 0))
)

# Indicador de pago tardío:
# identifica si el cliente pagó después de la fecha límite
# Luego puede agregarse para obtener proporción de pagos tardíos
installments_fe['LATE_PAYMENT'] = (
    installments_fe['DPD'] > 0
).astype(int)

# Indicador de pago adelantado:
# identifica si el cliente pagó antes de la fecha límite
# Puede reflejar mejor comportamiento financiero
installments_fe['EARLY_PAYMENT'] = (
    installments_fe['DBD'] > 0
).astype(int)

# Indicador de subpago:
# identifica si el cliente pagó menos de lo esperado
# Útil para detectar problemas de liquidez o incumplimiento parcial
installments_fe['UNDERPAID'] = (
    installments_fe['AMT_PAYMENT']
    < installments_fe['AMT_INSTALMENT']
).astype(int)

# Indicador de sobrepago:
# identifica pagos mayores al monto esperado
# Puede indicar liquidaciones anticipadas o pagos extraordinarios
installments_fe['OVERPAID'] = (
    installments_fe['AMT_PAYMENT']
    > installments_fe['AMT_INSTALMENT']
).astype(int)

# Indicador de pago exacto:
# identifica si el cliente pagó exactamente el monto esperado
# Ayuda a medir estabilidad y cumplimiento regular
installments_fe['EXACT_PAYMENT'] = (
    installments_fe['AMT_PAYMENT']
    == installments_fe['AMT_INSTALMENT']
).astype(int)


# Aggregacion por cliente

installments_summary = (
    installments_fe
    .groupby('SK_ID_CURR')
    .agg({

        # Número de créditos previos
        'SK_ID_PREV': ['nunique'],

        # Número de versiones
        'NUM_INSTALMENT_VERSION': ['nunique','max'],

        # Número de cuotas
        'NUM_INSTALMENT_NUMBER': ['max','mean','count'],

        # Fechas programadas
        'DAYS_INSTALMENT': [ 'min','max','mean'],

        # Fechas reales de pago
        'DAYS_ENTRY_PAYMENT': [ 'min','max','mean'],

        # Monto esperado
        'AMT_INSTALMENT': ['min','max','mean', 'sum'],

        # Monto pagado
        'AMT_PAYMENT': ['min','max','mean','sum'],

        # Diferencia de pago
        'PAYMENT_DIFF': ['min','max','mean','sum'],

        # Ratio de pago
        'PAYMENT_RATIO': ['min','max','mean'],

        # Retrasos
        'DPD': ['max','mean','sum'],

        # Adelantos
        'DBD': ['max','mean','sum'],

        # Indicadores
        'LATE_PAYMENT': ['mean','sum'],
        'EARLY_PAYMENT': ['mean','sum'],
        'UNDERPAID': ['mean','sum'],
        'OVERPAID': ['mean','sum'],
        'EXACT_PAYMENT': ['mean','sum']

    })
)


# Aplanar nombres de columnas

installments_summary.columns = [
    f'INSTAL_{col[0]}_{col[1]}'
    for col in installments_summary.columns
]

installments_summary = installments_summary.reset_index()

# Limpieza de infinitos

installments_summary = installments_summary.replace(
    [np.inf, -np.inf],
    np.nan
)

```

## pos_cash_balance

```python
pos_cash_balance.info()
```

```python
pos_cash_fe = pos_cash_balance.copy()

# Variables derivadas

# Ratio de cuotas restantes:
# mide qué proporción del crédito sigue pendiente
# Valores altos indican créditos aún poco avanzados
pos_cash_fe['INSTALMENT_FUTURE_RATIO'] = np.where(
    pos_cash_fe['CNT_INSTALMENT'] > 0,
    pos_cash_fe['CNT_INSTALMENT_FUTURE']
    / pos_cash_fe['CNT_INSTALMENT'],
    np.nan
)

# Cuotas ya pagadas:
# aproxima cuántas cuotas han sido completadas
pos_cash_fe['INSTALMENTS_PAID'] = (
    pos_cash_fe['CNT_INSTALMENT']
    - pos_cash_fe['CNT_INSTALMENT_FUTURE']
)

# Indicador de retraso:
# identifica si existe atraso en pagos POS/Cash
pos_cash_fe['HAS_DPD'] = (
    pos_cash_fe['SK_DPD'] > 0
).astype(int)

# Indicador de atraso severo:
# SK_DPD_DEF representa atraso con tolerancia vencida
# Suele capturar mora más grave.
pos_cash_fe['HAS_DPD_DEF'] = (
    pos_cash_fe['SK_DPD_DEF'] > 0
).astype(int)

# Indicador de contrato activo:
# identifica créditos POS/Cash aún vigentes
pos_cash_fe['CONTRACT_ACTIVE'] = (
    pos_cash_fe['NAME_CONTRACT_STATUS']
    == 'Active'
).astype(int)

# Indicador de contrato completado:
# identifica créditos finalizados correctamente
pos_cash_fe['CONTRACT_COMPLETED'] = (
    pos_cash_fe['NAME_CONTRACT_STATUS']
    == 'Completed'
).astype(int)

# Indicador de contrato firmado:
# puede representar créditos recientemente originados
pos_cash_fe['CONTRACT_SIGNED'] = (
    pos_cash_fe['NAME_CONTRACT_STATUS']
    == 'Signed'
).astype(int)

# Indicador de demanda/deuda problemática:
# contratos asociados a problemas legales o incumplimiento
pos_cash_fe['CONTRACT_DEMAND'] = (
    pos_cash_fe['NAME_CONTRACT_STATUS']
    == 'Demand'
).astype(int)

# Indicador de créditos aprobados:
# útil para medir historial de aprobación
pos_cash_fe['CONTRACT_APPROVED'] = (
    pos_cash_fe['NAME_CONTRACT_STATUS']
    == 'Approved'
).astype(int)

# One Hot Encoding
pos_cash_fe = pd.get_dummies(
    pos_cash_fe,
    columns=['NAME_CONTRACT_STATUS'],
    dtype=int
)

# Aggregation por cliente
pos_cash_summary = (
    pos_cash_fe
    .groupby('SK_ID_CURR')
    .agg({

        # Cantidad de créditos POS/CASH
        'SK_ID_PREV': ['nunique'],

        # Historial temporal
        'MONTHS_BALANCE': ['min','max','mean'],

        # Número total de cuotas
        'CNT_INSTALMENT': ['min','max','mean','sum'],

        # Cuotas futuras
        'CNT_INSTALMENT_FUTURE': ['min','max','mean','sum'],

        # Ratio cuotas restantes
        'INSTALMENT_FUTURE_RATIO': ['min','max','mean'],

        # Cuotas pagadas
        'INSTALMENTS_PAID': ['min','max','mean','sum'],

        # Days past due
        'SK_DPD': ['max','mean','sum'],

        # Days past due default
        'SK_DPD_DEF': ['max','mean','sum'],

        # Indicadores
        'HAS_DPD': ['mean','sum'],
        'HAS_DPD_DEF': ['mean','sum'],
        'CONTRACT_ACTIVE': ['mean','sum'],
        'CONTRACT_COMPLETED': ['mean','sum'],
        'CONTRACT_SIGNED': ['mean','sum'],
        'CONTRACT_DEMAND': ['mean','sum'],
        'CONTRACT_APPROVED': ['mean','sum']

    })
)

# Aplanar nombres de columnas
pos_cash_summary.columns = [
    f'POS_{col[0]}_{col[1]}'
    for col in pos_cash_summary.columns
]

pos_cash_summary = pos_cash_summary.reset_index()

# Limpieza
pos_cash_summary = pos_cash_summary.replace(
    [np.inf, -np.inf],
    np.nan
)
```
